```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Q1. Follow along with these steps: \
a) Create a figure object called fig using plt.figure() \
b) Use add_axes to add an axis to the figure canvas at [0,0,1,1]. Call this new axis ax.\
c) Plot (x,y) on that axes and set the labels and titles

```python
x = np.linspace(0, 100, 100)
y = 2 * x

fig = plt.figure()

ax = fig.add_axes([0, 0, 1, 1])

ax.plot(x, y, color='blue')

ax.set_title('title')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_xlim(0,max(x))
ax.set_ylim(0,max(y))

plt.show()
```

Q2. Create a figure object and put two axes on it, ax1 and ax2. Located at [0,0,1,1] and [0.2,0.5,.2,.2]
respectively.\
Now plot (x,y) on both axes. And call your figure object to show it.

```python
x = np.arange(1, 11)
y = x ** 2

fig = plt.figure()

ax1 = fig.add_axes([0, 0, 1, 1])
ax2 = fig.add_axes([0.2, 0.5, 0.2, 0.2])

ax1.plot(x, y)
ax1.set_title("Main Plot")
ax1.set_xlim(0,max(x))
ax1.set_ylim(0,max(y))

ax2.plot(x, y)
ax2.set_title("Inset Plot")
ax2.set_xlim(0,max(x))
ax2.set_ylim(0,max(y))

plt.show()
```

Q3. Use the company sales dataset csv file, read Total profit of all months and show it using a line plotTotal profit data provided for each month.\ Generated line plot must include the following properties: –\
a. X label name = Month Number\
b. Y label name = Total profit

```python
df = pd.read_csv('company-sales.csv')
df.plot(x='month_number', y='total_profit', marker='o')
plt.xlabel('Month Number')
plt.ylabel('Total profit')
plt.title('Total Profit by Month')
plt.grid(True)
plt.show()
```

Q4. Use the company sales dataset csv file, get total profit of all months and show line plot with the following\
Style properties. Generated line plot must include following Style properties: –\
a. Line Style dotted and Line-color should be red\
b. Show legend at the lower right location.\
c. X label name = Month Number\
d. Y label name = Sold units number\
e. Add a circle marker.\
f. Line marker color as read\
g. Line width should be 3

```python
df = pd.read_csv('company-sales.csv')

df.plot(
    x='month_number',
    y='total_profit',
    linestyle='dotted',
    color='red',
    marker='o',
    markerfacecolor='red',
    linewidth=3,
    label='Total Profit'
)
plt.xlabel('Month Number')
plt.ylabel('Sold units number')
plt.title('Total Profit by Month (Styled)')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

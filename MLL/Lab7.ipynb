Q1) Implement in python program of the following problems using Bayes Theorem. 
a)  Of  the  students  in  the  college,  60%  of  the  students  reside  in  the  hostel  and  40%  of  the  students  are  day 
scholars. Previous year results report that 30% of all students who stay in the hostel scored A Grade and 20% 
of day scholars scored A grade. At the end of the year, one student is chosen at random and found that he/she 
has an A grade. What is the probability that the student is a hosteler?  
b) Suppose you're testing for a rare disease, and you have the following information:  The disease has a prevalence of 0.01 (1% of the population has the disease).  The test is not perfect:  The test correctly identifies the disease (true positive) 99% of the time (sensitivity).  The test incorrectly indicates the disease (false positive) 2% of the time (1 - specificity). 
Calculate the probability of having the disease given a positive test result using Bayes' theorem. 

```python
p_hostel = 0.60
p_day_scholar = 0.40

p_A_hostel = 0.30
p_A_day_scholar = 0.20

p_A = (p_A_hostel * p_hostel) + (p_A_day_scholar * p_day_scholar)

p_hostel_A = (p_A_hostel * p_hostel) / p_A

print("Total probability of an A grade:", p_A)
print("Probability that the student is a hosteler given they have an A grade:", p_hostel_A)
print("Percentage:", p_hostel_A * 100)
```

```python
p_disease = 0.01
p_no_disease = 0.99

p_positive_disease = 0.99
p_positive_no_disease = 0.02

p_positive = (
    p_positive_disease * p_disease
    + p_positive_no_disease * p_no_disease
)

p_disease_positive = (
    p_positive_disease * p_disease
) / p_positive

print("Total probability of testing positive:", p_positive)
print("Probability of having the disease given a positive test:", p_disease_positive)
print("Percentage:", p_disease_positive * 100)
```

Q2)Develop a function python code for Naïve Bayes classifier from scratch without using scikit-learn library, 
to predict whether the buyer should buy computer or not. Consider a following sample training dataset stored 
in a CSV file containing information about following buyer conditions (such as “<=30,” “medium,” “Yes,” 
and “fair”) and whether the player played golf (“Yes” or “No”). 

```python
def naive_bayes(data, test_data):
    classes = data["buys_computer"].unique()

    probabilities = {}

    for current_class in classes:
        class_data = data[data["buys_computer"] == current_class]

        prior = len(class_data) / len(data)

        probability = prior

        for feature in test_data:
            count = len(
                class_data[class_data[feature] == test_data[feature]]
            )

            conditional_probability = count / len(class_data)

            probability = probability * conditional_probability

        probabilities[current_class] = probability

    prediction = max(probabilities, key=probabilities.get)

    return prediction, probabilities


data = pd.read_csv("buys_computer.csv")

test_data = {
    "age": "<=30",
    "income": "medium",
    "student": "yes",
    "credit_rating": "fair"
}

prediction, probabilities = naive_bayes(data, test_data)

print("Probability of No:", probabilities["no"])
print("Probability of Yes:", probabilities["yes"])
print("Prediction:", prediction)
```

Q3)  Write  a  Python  function  to  implement  the  Naive  Bayes  classifier  without  using  the  scikit-learn  library  for  the 
following sample training dataset stored as a .CSV file.  
a. Build a classifier that determines whether a text is about sports or not. \
b. Determine which tag the sentence "A very close game" belongs to. 

```python
def tokenize(text):
    return re.findall(r'\b\w+\b', text.lower())


def train_naive_bayes(data):
    classes = data["Tag"].unique()

    word_counts = {}
    total_words = {}
    class_counts = {}
    vocabulary = set()

    for current_class in classes:
        class_data = data[data["Tag"] == current_class]

        words = []

        for text in class_data["Text"]:
            words.extend(tokenize(text))

        word_counts[current_class] = Counter(words)
        total_words[current_class] = len(words)
        class_counts[current_class] = len(class_data)

        vocabulary.update(words)

    return word_counts, total_words, class_counts, vocabulary


def predict(text, data):
    word_counts, total_words, class_counts, vocabulary = train_naive_bayes(data)

    words = tokenize(text)

    probabilities = {}

    total_documents = len(data)

    for current_class in class_counts:
        prior = class_counts[current_class] / total_documents

        probability = prior

        for word in words:
            count = word_counts[current_class][word]

            conditional_probability = (
                (count + 1)
                / (total_words[current_class] + len(vocabulary))
            )

            probability = probability * conditional_probability

        probabilities[current_class] = probability

    prediction = max(probabilities, key=probabilities.get)

    return prediction, probabilities


data = pd.read_csv("sports_text.csv")

test_sentence = "A very close game"

prediction, probabilities = predict(test_sentence, data)

print("Target Sentence:", test_sentence)
print()
print("--- Naive Bayes Probability Scores ---")

for current_class in probabilities:
    print(
        "Score for '" + current_class + "':",
        probabilities[current_class]
    )

print("--------------------------------------")
print("Final Prediction:", prediction)
```
